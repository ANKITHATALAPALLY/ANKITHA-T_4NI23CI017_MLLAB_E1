```python
import pandas as pd

data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High',    'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High',    'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High',    'Strong', 'Cool', 'Change', 'Yes']
]

df = pd.DataFrame(data, columns=[
    'Sky', 'Temperature', 'Humidity',
    'Wind', 'Water', 'Forecast', 'EnjoySport'
])

X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values

# Initial Specific hypothesis
S = ['0'] * len(X[0])

# Initial General hypothesis
G = [['?'] * len(X[0])]

# Candidate Elimination
for i in range(len(X)):
    if y[i] == 'Yes':
        # Remove inconsistent hypotheses from G
        G = [g for g in G if all(g[j] == '?' or g[j] == X[i][j] for j in range(len(X[i])))]

        # Generalize S
        for j in range(len(X[i])):
            if S[j] == '0':
                S[j] = X[i][j]
            elif S[j] != X[i][j]:
                S[j] = '?'

    else:
        # Remove inconsistent hypotheses from S
        for j in range(len(X[i])):
            if S[j] == X[i][j]:
                S[j] = '?'

        # Keep G general for this simple example
        G = [g for g in G if any(g[j] != X[i][j] for j in range(len(X[i])))]

# Display results
print("Final Specific Hypothesis (S):")
print(S)

print("\nFinal General Hypothesis (G):")
for g in G:
    print(g)
```
